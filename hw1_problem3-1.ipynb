```python
import json
import numpy as np
import torch
from torch import nn
```

```python
rng = np.random.default_rng(42)
params = {
    "W1": rng.uniform(-1, 1, (3, 2)).round(3).tolist(),
    "b1": rng.uniform(-0.5, 0.5, 3).round(3).tolist(),
    "W2": rng.uniform(-1, 1, (1, 3)).round(3).tolist(),
    "b2": rng.uniform(-0.5, 0.5, 1).round(3).tolist(),
}
with open("problem3_weights.json", "w") as f:
    json.dump({"parameters": params}, f)
```

```python
class FixedNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden = nn.Linear(2, 3, dtype=torch.float64)
        self.output = nn.Linear(3, 1, dtype=torch.float64)

    def forward(self, x):
        return self.output(torch.relu(self.hidden(x)))  # logits for BCEWithLogitsLoss

    def predict_proba(self, x):
        return torch.sigmoid(self(x))
```

```python
with open("problem3_weights.json") as f:
    params = json.load(f)["parameters"]

model = FixedNet()
model.load_state_dict({
    "hidden.weight": torch.tensor(params["W1"], dtype=torch.float64),
    "hidden.bias": torch.tensor(params["b1"], dtype=torch.float64),
    "output.weight": torch.tensor(params["W2"], dtype=torch.float64),
    "output.bias": torch.tensor(params["b2"], dtype=torch.float64),
});
```
